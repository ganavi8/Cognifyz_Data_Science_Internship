# Level 1 - Task 2: Descriptive Analysis

### Objective

Perform descriptive analysis on the Zomato restaurant dataset by calculating
statistical measures for numerical variables and analyzing categorical
variables such as Country Code, City, and Cuisines.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Outputs/zomato_cleaned.csv")

print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Numerical Columns:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical Columns:")
print(df.select_dtypes(include="object").columns.tolist())
```

# Numerical descriptive statistics

```python
numerical_columns = df.select_dtypes(include=np.number).columns

descriptive_stats = df[numerical_columns].describe()

descriptive_stats
```

```python
print("Mean:")
print(df[numerical_columns].mean())
```

```python
print("Median:")
print(df[numerical_columns].median())
```

```python
print("Standard Deviation:")
print(df[numerical_columns].std())
```

 # Country analysis

```python
country_counts = df["Country Code"].value_counts()

print("Restaurant Count by Country:")
print(country_counts)
```

```python
plt.figure(figsize=(10, 6))

country_counts.head(15).plot(kind="bar")

plt.title("Restaurant Distribution by Country")
plt.xlabel("Country Code")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# City analysis

```python
city_counts = df["City"].value_counts()

print("Top 20 Cities:")
print(city_counts.head(20))
```

```python
plt.figure(figsize=(12, 7))

city_counts.head(15).plot(kind="bar")

plt.title("Top 15 Cities by Number of Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=60)

plt.tight_layout()
plt.show()
```

 # Cuisine analysis

```python
top_cuisines = df["Cuisines"].value_counts()

print("Top 20 Cuisine Combinations:")
print(top_cuisines.head(20))
```

```python
cuisine_list = (
    df["Cuisines"]
    .dropna()
    .str.split(", ")
    .explode()
)

cuisine_counts = cuisine_list.value_counts()

print("Top 20 Individual Cuisines:")
print(cuisine_counts.head(20))
```

```python
plt.figure(figsize=(12, 7))

cuisine_counts.head(15).plot(kind="bar")

plt.title("Top 15 Individual Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=60)

plt.tight_layout()
plt.show()
```

# Identify highest city and cuisine

```python
top_city = city_counts.idxmax()
top_city_count = city_counts.max()

print("City with the highest number of restaurants:", top_city)
print("Number of restaurants:", top_city_count)
```

```python
top_cuisine = cuisine_counts.idxmax()
top_cuisine_count = cuisine_counts.max()

print("Most popular cuisine:", top_cuisine)
print("Number of restaurants:", top_cuisine_count)
```

 # outputs

```python
descriptive_stats.to_csv(
    "../Outputs/descriptive_statistics.csv"
)

city_counts.head(20).to_csv(
    "../Outputs/top_cities.csv",
    header=["Restaurant Count"]
)

cuisine_counts.head(20).to_csv(
    "../Outputs/top_cuisines.csv",
    header=["Restaurant Count"]
)

print("All Task 2 outputs saved successfully.")
```
