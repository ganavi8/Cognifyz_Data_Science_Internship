# Level 2 - Task 3: Feature Engineering

## Objective
Perform feature engineering on the Zomato restaurant dataset by creating
new meaningful features from existing columns.

### Tasks
- Extract restaurant name length
- Extract address length
- Encode table booking availability
- Encode online delivery availability
- Create additional useful numerical features
- Save the feature-engineered dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Outputs/zomato_cleaned.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
required_columns = [
    "Restaurant Name",
    "Address",
    "Has Table booking",
    "Has Online delivery"
]

missing_columns = [col for col in required_columns if col not in df.columns]

if missing_columns:
    print("Missing columns:", missing_columns)
else:
    print("All required columns are available.")
    
```

 # Restaurant Name Length

```python
df["Restaurant Name Length"] = (
    df["Restaurant Name"]
    .fillna("")
    .astype(str)
    .str.len()
)

print("Restaurant Name Length created successfully.")
df[["Restaurant Name", "Restaurant Name Length"]].head()
```

 # Address Length

```python
df["Address Length"] = (
    df["Address"]
    .fillna("")
    .astype(str)
    .str.len()
)

print("Address Length created successfully.")
df[["Address", "Address Length"]].head()
```

 # Encode Table Booking

```python
df["Has Table Booking Encoded"] = (
    df["Has Table booking"]
    .map({"Yes": 1, "No": 0})
)

print("Table Booking encoding:")
print(df["Has Table Booking Encoded"].value_counts(dropna=False))
```

 # Encode Online Delivery

```python
df["Has Online Delivery Encoded"] = (
    df["Has Online delivery"]
    .map({"Yes": 1, "No": 0})
)

print("Online Delivery encoding:")
print(df["Has Online Delivery Encoded"].value_counts(dropna=False))
```

# Create Price Range Feature

```python
if "Price range" in df.columns:
    df["Price Range Category"] = df["Price range"].map({
        1: "Low",
        2: "Medium",
        3: "High",
        4: "Very High"
    })

print(df[["Price range", "Price Range Category"]].head())
```

# Create Votes Log Feature

```python
if "Votes" in df.columns:
    df["Log Votes"] = np.log1p(df["Votes"])

print(df[["Votes", "Log Votes"]].head())
```

# Check Engineered Features

```python
engineered_features = [
    "Restaurant Name Length",
    "Address Length",
    "Has Table Booking Encoded",
    "Has Online Delivery Encoded",
    "Price Range Category",
    "Log Votes"
]

available_features = [
    col for col in engineered_features
    if col in df.columns
]

print("Engineered Features:")
print(available_features)

df[available_features].head(10)
```

# Dataset Information

```python
print("Final Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df[available_features].isnull().sum())
```

# Feature-Engineered Dataset

```python
output_path = "../Outputs/feature_engineered_zomato.csv"

df.to_csv(output_path, index=False)

print(f"Feature-engineered dataset saved to: {output_path}")
```

```python
df_check = pd.read_csv("../Outputs/feature_engineered_zomato.csv")

print("Saved Dataset Shape:", df_check.shape)

print("\nNew Features:")
print([
    col for col in df_check.columns
    if col in engineered_features
])

df_check.head()
```
